```python
import pandas as pd

# Load the sales data
df = pd.read_csv('sales_data.csv')

# Display the first few rows to confirm loading and inspect data types
print("Original DataFrame head:")
display(df.head())

# Display initial info to see data types
print("\nOriginal DataFrame info:")
df.info()
```

### Cleaning `Total_Sale` and `Unit_Price` columns

As observed from the `df.info()` output or by looking at the raw data, the `Total_Sale` and `Unit_Price` columns are likely objects (strings) because of the '$' sign. We need to remove the '$' and convert them to numeric (float) type for calculations.

```python
# Clean 'Total_Sale' column: remove '$' and convert to numeric
df['Total_Sale'] = df['Total_Sale'].astype(str).str.replace('$', '', regex=False).astype(float)

# Also clean 'Unit_Price' for consistency, though not strictly needed for this task
df['Unit_Price'] = df['Unit_Price'].astype(str).str.replace('$', '', regex=False).astype(float)

# Verify the data types after cleaning
print("DataFrame info after cleaning 'Total_Sale' and 'Unit_Price':")
df.info()
```

### Task 1: Total sales by region, highest first

```python
# Calculate total sales by region, highest first
total_sales_by_region = df.groupby('Region')['Total_Sale'].sum().sort_values(ascending=False)

# Display the result
print("Total Sales by Region (highest first):")
display(total_sales_by_region)
```

### Cleaning `Quantity` column

Based on initial observation, the `Quantity` column also contains '$' signs, making it an `object` type. We need to clean this column by removing the '$' and converting it to an integer for accurate unit counting.

```python
# Clean 'Quantity' column: remove '$' and convert to integer
df['Quantity'] = df['Quantity'].astype(str).str.replace('$', '', regex=False).astype(int)

# Verify the data types after cleaning
print("DataFrame info after cleaning 'Quantity':")
df.info()
```

```python
print('\nRe-running cell 20728af0 for Task 2: Total units sold per product:')
# Calculate total units sold per product
units_sold_per_product = df.groupby('Product')['Quantity'].sum().sort_values(ascending=False)

# Display the result
print("Total Units Sold per Product (highest first):")
display(units_sold_per_product)
```

```python
print('\nRe-running cell 49a3eef4 for Task 3: Sales Rep Revenue:')
# Identify which sales rep brought in the most revenue
sales_rep_revenue = df.groupby('Sales_Rep')['Total_Sale'].sum().sort_values(ascending=False)

# Display the result
print("Sales Revenue by Sales Representative (highest first):")
display(sales_rep_revenue)
```
